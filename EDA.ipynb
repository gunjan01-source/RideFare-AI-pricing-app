```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("dynamic_pricing_data.csv")
print(df.head())
```

```python
null_values = df.isna().sum()
print("null_values:\n",null_values)
```

```python
duplicated_values = df.duplicated().sum()
print("duplicated_values:\n",duplicated_values)
```

```python
print(df["Season"].unique())
print(df["DayTime"].unique())
```

```python
plt.figure(figsize=(8,5))

sns.regplot(
    x="Demand",
    y="Price",
    data=df,
    scatter_kws={"alpha":0.4},
    line_kws={"color":"#00008B"},
    color="#0BA3FF"
)

plt.title("Demand vs Price")
plt.xlabel("Demand")
plt.ylabel("Price")

plt.grid(alpha=0.3)

plt.show()
```

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(10,8))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="Blues",
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(
    x="Season",
    y="Price",
    data=df,
    color="#0BA3FF"
)
plt.title("Price Distribution by Season")
plt.xlabel("Season")
plt.ylabel("Price")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(
    x="DayTime",
    y="Price",
    data=df,
    color="#0BA3FF"
)
plt.title("Price Distribution by Season")
plt.xlabel("Season")
plt.ylabel("Price")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.histplot(df["Price"], bins=30, kde=True)
plt.title("Distribution of Price")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

```python
print(df["Price"].skew())
```

Price Distribution: The target variable (Price) has a skewness of 0.28, indicating an approximately symmetric distribution with a slight right skew. Most rides are concentrated around the middle price range, while a small number of higher-priced rides occur during high-demand scenarios such as festivals and peak hours. Since the skewness is low, no logarithmic transformation of the target variable is required before model training.

```python
ratio = df["Demand"] / df["Stock"]

plt.figure(figsize=(7,5))

sns.scatterplot(
    x=ratio,
    y=df["Price"],
    alpha=0.4
)

plt.title("Price vs Demand/Stock Ratio")
plt.xlabel("Demand / Stock Ratio")
plt.ylabel("Price")

plt.show()
```

```python
print(df.corr(numeric_only=True)["Price"].sort_values(ascending=False))
```

#Exploratory Data Analysis Summary
The dataset contains 20,000 rows with no missing values or duplicate records.
Numerical features were explored using summary statistics and correlation analysis.
Demand, CompetitorPrice, Distance_km, and HistoricalSales show positive relationships with Price.
Stock shows a negative relationship with Price, consistent with dynamic pricing.
Festival and peak-hour periods generally have higher ride prices than Normal and Exam periods.
The target variable (Price) has a skewness of 0.28, indicating an approximately symmetric distribution, so no log transformation was applied.
Overall, the dataset appears clean and suitable for supervised regression modeling.
